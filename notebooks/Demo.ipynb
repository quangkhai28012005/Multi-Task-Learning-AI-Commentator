# Multi-Task Learning AI Commentator - Demo Notebook

This notebook demonstrates the capabilities of the Multi-Task Learning AI Commentator system for analyzing football match events.

```python
# Import necessary libraries
import sys
sys.path.insert(0, '..')

from src.inference.commentator import AICommentator
from src.data_processing.data_loader import DataLoader
import pandas as pd
import json
```

## 1. Initialize the AI Commentator

```python
# Initialize commentator
print('Initializing AI Commentator...')
commentator = AICommentator()
print('✓ Commentator initialized!')
print(f'Device: {commentator.device}')
```

## 2. Load Sample Data

```python
# Load sample events
sample_data = DataLoader.load_json('../data/samples/sample_data.json')
print(f'Loaded {len(sample_data)} sample events')

# Display first event
print('\nFirst Event:')
print(json.dumps(sample_data[0], ensure_ascii=False, indent=2))
```

## 3. Analyze Single Event

```python
# Analyze a single event
event = sample_data[0]
result = commentator.analyze(event)

print(f'Event: {event["description"]}\n')
print(f'AI Analysis:')
print(f'  Action Type: {result["action_type"]}')
print(f'  Commentary: {result["commentary"]}')
print(f'  Sentiment: {result["sentiment"]}')
print(f'  Importance: {result["importance_score"]} / 10')
print(f'  Confidence: {result["confidence"]*100:.1f}%')
```

## 4. Batch Analysis

```python
# Batch analyze all events
results = commentator.batch_analyze(sample_data)

# Create results dataframe
df_results = pd.DataFrame([
    {
        'Match': event.get('match_id'),
        'Time': event.get('timestamp'),
        'Team': result['team'],
        'Player': result['player'],
        'Action': result['action_type'],
        'Sentiment': result['sentiment'],
        'Importance': result['importance_score'],
        'Confidence': round(result['confidence']*100, 1)
    }
    for event, result in zip(sample_data, results)
])

print('Analysis Results:')
print(df_results.to_string(index=False))
```

## 5. Analyze Custom Event

```python
# Create and analyze custom event
custom_event = {
    'action_type': 'goal',
    'team': 'Manchester United',
    'player': 'Cristiano Ronaldo',
    'description': 'Sút xa chính xác vào góc tường, thủ môn không có cơ hội cứu thua'
}

custom_result = commentator.analyze(custom_event)

print('Custom Event Analysis:')
print(f'Event: {custom_event["description"]}\n')
print(f'Results:')
print(json.dumps(custom_result, ensure_ascii=False, indent=2))
```

## 6. Sentiment Distribution Analysis

```python
# Analyze sentiment distribution
sentiments = [result['sentiment'] for result in results]
sentiment_counts = pd.Series(sentiments).value_counts()

print('Sentiment Distribution:')
print(sentiment_counts)
print(f'\nTotal Events: {len(results)}')
```

## 7. Action Type Distribution

```python
# Analyze action distribution
actions = [result['action_type'] for result in results]
action_counts = pd.Series(actions).value_counts()

print('Action Type Distribution:')
print(action_counts)
```

## 8. Average Importance Score

```python
# Calculate average importance
importance_scores = [result['importance_score'] for result in results]
avg_importance = sum(importance_scores) / len(importance_scores)

print(f'Average Importance Score: {avg_importance:.2f} / 10')
print(f'Min Importance: {min(importance_scores):.2f}')
print(f'Max Importance: {max(importance_scores):.2f}')
```
